```python
import pandas as pd

transactions = pd.read_csv("Data/Raw/08_investor_transactions.csv")

print(transactions.head())
print(transactions.columns.tolist())
```

```python
plt.figure(figsize=(8,5))

transactions['transaction_type'].value_counts().plot(
    kind='bar'
)

plt.title("Transaction Type Distribution")
plt.xlabel("Transaction Type")
plt.ylabel("Count")

plt.tight_layout()

plt.savefig("reports/charts/chart09_transaction_type_distribution.png")

plt.show()
```

```python
plt.figure(figsize=(8,5))

transactions['age_group'].value_counts().plot(
    kind='bar'
)

plt.title("Investor Age Group Distribution")
plt.xlabel("Age Group")
plt.ylabel("Count")

plt.tight_layout()

plt.savefig("reports/charts/chart12_age_group_distribution.png")

plt.show()
```

```python
plt.figure(figsize=(10,5))

transactions['state'].value_counts().head(10).plot(
    kind='bar'
)

plt.title("Top 10 States by Transaction Count")
plt.xlabel("State")
plt.ylabel("Count")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig("reports/charts/chart13_state_distribution.png")

plt.show()
```

```python
transactions['transaction_date'] = pd.to_datetime(
    transactions['transaction_date']
)

daily_txn = transactions.groupby(
    transactions['transaction_date'].dt.date
).size()

plt.figure(figsize=(12,5))

plt.plot(daily_txn.index, daily_txn.values)

plt.title("Daily Transaction Trend")
plt.xlabel("Date")
plt.ylabel("Transactions")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig("reports/charts/chart14_daily_transaction_trend.png")

plt.show()
```

```python
plt.figure(figsize=(7,7))

transactions['transaction_type'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%'
)

plt.title("Buy vs Sell Transaction Share")
plt.ylabel("")

plt.tight_layout()

plt.savefig("reports/charts/chart10_buy_sell_pie.png")

plt.show()
```

```python
import matplotlib.pyplot as plt
import os

os.makedirs("Reports/charts", exist_ok=True)

# Convert to numeric if needed
performance['return_3yr_pct'] = pd.to_numeric(
    performance['return_3yr_pct'],
    errors='coerce'
)

top = performance.nlargest(10, 'return_3yr_pct')

plt.figure(figsize=(10,5))

plt.barh(top['scheme_name'], top['return_3yr_pct'])

plt.title("Top 10 Schemes by 3-Year Return")
plt.xlabel("3-Year Return (%)")

plt.tight_layout()

plt.savefig("Reports/charts/chart11_top_schemes.png")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# Load AUM dataset
aum = pd.read_csv("Data/Raw/03_aum_by_fund_house.csv")

# Create charts folder if it doesn't exist
os.makedirs("Reports/charts", exist_ok=True)

# Convert last column to numeric
aum[aum.columns[-1]] = pd.to_numeric(
    aum[aum.columns[-1]],
    errors='coerce'
)

# Top 5 fund houses by AUM
top = aum.sort_values(
    by=aum.columns[-1],
    ascending=False
).head(5)

# Create pie chart
plt.figure(figsize=(7,7))

plt.pie(
    top.iloc[:, -1],
    labels=top.iloc[:, 0],
    autopct='%1.1f%%'
)

plt.title("Top 5 AUM Share")

plt.savefig("Reports/charts/chart15_aum_pie.png")

plt.show()
```
